```python
data = data.drop(columns=['salary_range'])
```

```python
text_columns = [
    'department',
    'company_profile',
    'requirements',
    'benefits',
    'employment_type',
    'required_experience',
    'required_education',
    'industry',
    'function',
    'location'
]

for col in text_columns:
    data[col] = data[col].fillna('Not Available')
```

```python
data['description'] = data['description'].fillna('Not Available')
```
